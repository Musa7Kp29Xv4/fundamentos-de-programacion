# TEMA 1. ALGORITMOS Y MODELO ENTRADA-PROCESO-SALIDA (EPS)

Un algoritmo es una serie de pasos ordenados que permiten resolver un problema.

El modelo Entrada-Proceso-Salida permite identificar qué información necesita un programa, qué operaciones debe realizar y qué resultado debe mostrar.

### Ejemplo aplicado a RUCAM

**Entrada:**
- Precio del producto.
- Cantidad que desea comprar el cliente.

**Proceso:**
- Multiplicar el precio por la cantidad.
- Calcular el importe de la compra.

**Salida:**
- Mostrar el producto.
- Mostrar la cantidad.
- Mostrar el total a pagar.

```python
# Datos de entrada
precio_articulo = 46
cantidad_articulo = 2

# Proceso
importe_compra = precio_articulo * cantidad_articulo

# Salida
print("SISTEMA RUCAM")
print("Precio por kilo de huevo: $", precio_articulo)
print("Cantidad comprada:", cantidad_articulo)
print("Total de la compra: $", importe_compra)
```

# TEMA 2. PSEINT Y PSEUDOCÓDIGO

El pseudocódigo permite representar los pasos de un algoritmo utilizando instrucciones sencillas, sin depender de un lenguaje de programación específico.

### Ejemplo

El siguiente algoritmo permite calcular el importe de una compra en RUCAM.

```python
# Simulación del algoritmo de compra

costo_producto = 25
unidades_compradas = 3

importe_final = costo_producto * unidades_compradas

print("Producto: Queso Oaxaca")
print("Unidades:", unidades_compradas)
print("Importe total: $", importe_final)
```

# TEMA 3. VARIABLES, TIPOS DE DATOS Y OPERADORES

Las variables permiten almacenar información que puede utilizarse durante la ejecución de un programa.

En Python existen diferentes tipos de datos:

- int: números enteros.
- float: números decimales.
- str: cadenas de texto.
- bool: valores verdadero o falso.
- list: listas que almacenan varios elementos.

Los operadores permiten realizar operaciones matemáticas, comparaciones y operaciones lógicas.

```python
# Declaración de variables

nombre_articulo = "Sabritas"
precio_unitario = 22
cantidad_solicitada = 4
descuento_aplicado = 0.0
producto_disponible = True

catalogo_tienda = ["Maruchan", "Huevo", "Queso Oaxaca", "Sabritas"]

# Mostrar información
print("Producto:", nombre_articulo)
print("Precio:", precio_unitario)
print("Cantidad:", cantidad_solicitada)
print("Descuento:", descuento_aplicado)
print("Disponible:", producto_disponible)
print("Catálogo:", catalogo_tienda)

# Identificar los tipos de datos
print("\nTipos de datos:")
print(type(nombre_articulo))
print(type(precio_unitario))
print(type(descuento_aplicado))
print(type(producto_disponible))
print(type(catalogo_tienda))
```

```python
# Operadores matemáticos
costo_unitario = 20
piezas = 5

suma = costo_unitario + piezas
resta = costo_unitario - piezas
multiplicacion = costo_unitario * piezas
division = costo_unitario / piezas

print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)

# Operadores relacionales
print("\nComparaciones:")
print(costo_unitario > piezas)
print(costo_unitario == piezas)
print(costo_unitario != piezas)

# Operadores lógicos
hay_existencias = True
venta_activa = True

print("\nOperaciones lógicas:")
print(hay_existencias and venta_activa)
print(hay_existencias or venta_activa)
print(not hay_existencias)
```

# TEMA 4. ENTRADAS Y SALIDAS SIMPLES

La instrucción input() permite solicitar información al usuario, mientras que print() permite mostrar mensajes y resultados.

Cuando se reciben datos numéricos mediante input(), es necesario convertirlos utilizando int() o float(), dependiendo del tipo de dato que se necesite.

```python
# Captura de información del cliente

cliente_tienda = input("Ingresa el nombre del cliente: ")
producto_elegido = input("Ingresa el producto que desea comprar: ")
valor_producto = float(input("Ingresa el precio del producto: "))
piezas_elegidas = int(input("Ingresa la cantidad: "))

# Cálculo
monto_a_pagar = valor_producto * piezas_elegidas

# Mostrar resultado
print("\nRESUMEN DE COMPRA")
print("Cliente:", cliente_tienda)
print("Producto:", producto_elegido)
print("Cantidad:", piezas_elegidas)
print("Total: $", monto_a_pagar)
```

# TEMA 5. ESTRUCTURAS DE DECISIÓN

Las estructuras condicionales permiten que un programa tome decisiones dependiendo de si se cumple o no una condición.

En Python se utilizan:

- if: ejecuta instrucciones cuando se cumple una condición.
- elif: permite evaluar otra condición.
- else: se ejecuta cuando ninguna de las condiciones anteriores se cumple.

### Ejemplo aplicado a RUCAM

Se utilizarán estas estructuras para identificar si el cliente puede pagar en efectivo o con tarjeta.

```python
# Selección de forma de pago

forma_pago_cliente = 2

if forma_pago_cliente == 1:
    print("El cliente pagará en efectivo.")

elif forma_pago_cliente == 2:
    print("El cliente pagará con tarjeta bancaria.")

else:
    print("La forma de pago seleccionada no es válida.")
```

# TEMA 6. ESTRUCTURA DE REPETICIÓN WHILE

El ciclo while permite repetir un conjunto de instrucciones mientras una condición sea verdadera.

Es útil cuando no conocemos exactamente cuántas veces se repetirá una operación.

### Ejemplo aplicado a RUCAM

El siguiente programa permite registrar varias compras y acumular el total de las ventas hasta que el usuario decida terminar.

```python
# Registro de compras utilizando while

acumulado_ventas = 0
seguir_registrando = "si"

while seguir_registrando == "si":

    importe_venta = float(input("Ingresa el importe de la venta: "))

    if importe_venta > 0:
        acumulado_ventas += importe_venta
        print("Venta agregada correctamente.")
    else:
        print("El importe debe ser mayor que cero.")

    seguir_registrando = input("¿Deseas registrar otra venta? (si/no): ").lower()

print("\nRESUMEN DEL DÍA")
print("Total acumulado: $", acumulado_ventas)
```

# TEMA 7. DEBUGGING Y DEPURACIÓN CON PDB

La depuración permite encontrar y corregir errores dentro de un programa.

Python cuenta con la herramienta PDB, que permite ejecutar el código paso a paso, revisar el contenido de las variables y localizar dónde ocurre un problema.

Algunos comandos de PDB son:

- n: ejecuta la siguiente línea.
- s: entra en una función.
- p: muestra el valor de una variable.
- c: continúa la ejecución.
- q: termina la sesión de depuración.

### Ejemplo aplicado a RUCAM

Se revisará el cálculo del total de una compra para comprobar que las operaciones se realicen correctamente.

```python
# Programa para revisar el cálculo de una compra

precio_revision = 25
cantidad_revision = 4

total_revision = precio_revision * cantidad_revision

print("Precio:", precio_revision)
print("Cantidad:", cantidad_revision)
print("Total calculado:", total_revision)
```

```python
import pdb

def calcular_compra(precio, cantidad):
    subtotal = precio * cantidad
    return subtotal

precio_prueba = 22
cantidad_prueba = 3

pdb.set_trace()

resultado_prueba = calcular_compra(precio_prueba, cantidad_prueba)

print("Total de la compra:", resultado_prueba)
```

# TEMA 8. ESTRUCTURA DE REPETICIÓN FOR

El ciclo for permite recorrer elementos de una lista o una secuencia y ejecutar instrucciones para cada elemento.

Es útil cuando conocemos los elementos que queremos recorrer.

### Ejemplo aplicado a RUCAM

Se utilizará un ciclo for para mostrar los productos que se encuentran disponibles en el catálogo de la tienda.

```python
# Catálogo de productos RUCAM

lista_articulos = [
    "Maruchan",
    "Queso Oaxaca",
    "Huevo por kilo",
    "Sabritas",
    "Rollo de papel suelto",
    "Cigarros sueltos"
]

print("CATÁLOGO DE RUCAM")

for posicion, articulo in enumerate(lista_articulos, start=1):
    print(posicion, "-", articulo)
```

# PROBLEMA INTEGRADOR 1. CÁLCULO DE UNA COMPRA

### Situación

En RUCAM se necesita calcular el importe de una compra realizada por un cliente.

El programa debe solicitar el precio y la cantidad de productos, calcular el subtotal y mostrar el total que debe pagar.

### Temas aplicados
- Variables y tipos de datos.
- Entrada y salida de información.
- Operadores matemáticos.
- Estructuras de decisión.

```python
# Cálculo de compra

precio_compra = float(input("Precio del producto: "))
cantidad_compra = int(input("Cantidad comprada: "))

subtotal_compra = precio_compra * cantidad_compra

if cantidad_compra > 0 and precio_compra > 0:
    print("\nResumen de compra")
    print("Subtotal: $", subtotal_compra)
else:
    print("Los datos ingresados no son válidos.")
```

# PROBLEMA INTEGRADOR 2. REGISTRO DE VENTAS

### Situación

RUCAM necesita acumular el importe de varias ventas realizadas durante el día.

El programa permitirá registrar diferentes importes hasta que el usuario decida terminar.

### Temas aplicados
- Variables.
- Entrada y salida.
- Operadores.
- Estructuras condicionales.
- Ciclo while.
- Acumuladores.

```python
# Acumulación de ventas

suma_del_dia = 0
numero_operaciones = 0

respuesta_usuario = "si"

while respuesta_usuario == "si":

    valor_venta = float(input("Importe de la venta: "))

    if valor_venta > 0:
        suma_del_dia += valor_venta
        numero_operaciones += 1
        print("Venta registrada.")
    else:
        print("El importe no es válido.")

    respuesta_usuario = input("¿Registrar otra venta? (si/no): ").lower()

print("\nRESUMEN DE VENTAS")
print("Ventas registradas:", numero_operaciones)
print("Total vendido: $", suma_del_dia)
```

# PROBLEMA INTEGRADOR 3. CATÁLOGO Y CÁLCULO DE IMPORTES

### Situación

En RUCAM se necesita mostrar los productos disponibles y calcular el importe correspondiente a cada uno según la cantidad registrada.

### Temas aplicados
- Listas.
- Variables.
- Operadores matemáticos.
- Ciclo for.
- Acumuladores.

```python
# Productos y precios de RUCAM

articulos_tienda = [
    "Maruchan",
    "Queso Oaxaca",
    "Huevo por kilo",
    "Sabritas",
    "Rollo de papel suelto",
    "Cigarros sueltos"
]

costos_tienda = [20, 25, 46, 22, 14, 8]

unidades_tienda = [2, 1, 1, 3, 2, 5]

total_general = 0

print("RESUMEN DE PRODUCTOS")

for indice in range(len(articulos_tienda)):

    importe_articulo = costos_tienda[indice] * unidades_tienda[indice]

    total_general += importe_articulo

    print(
        articulos_tienda[indice],
        "- Cantidad:",
        unidades_tienda[indice],
        "- Importe: $",
        importe_articulo
    )

print("\nTotal de todos los productos: $", total_general)
```
