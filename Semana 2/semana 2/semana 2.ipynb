```python
# ==========================================
# ACTIVIDAD 2: LA TAQUILLA DEL MUSEO
# ==========================================

# 1. DEFINICIÓN DE CONSTANTES (Precios y Descuentos)
PRECIO_NINO_MENOR = 0.0        # Menos de 3 años[cite: 3]
PRECIO_MENOR = 30.0             # De 3 a 17 años[cite: 3]
PRECIO_ADULTO = 45.0            # 18 años o más[cite: 3]

DESC_ADULTO_MAYOR = 0.12        # 12% para 60 años o más[cite: 3]
DESC_ESTUDIANTE_PROFESOR = 0.10 # 10% para estudiantes o profesores[cite: 3]

# 2. INICIALIZACIÓN DE CONTADORES Y ACUMULADORES GLOBALES
visitantes_cobrados = 0
total_descuentos = 0.0
total_pagar = 0.0

print("=== TAQUILLA DEL MUSEO ===")

# 3. CICLO PRINCIPAL DE ATENCIÓN
while True:
    # Preguntar cuántos boletos se van a comprar
    while True:
        entrada = input("\n¿Cuántos boletos desea comprar? (escribir 'fin' para cerrar la taquilla): ").strip().lower()
        
        # Validación de salida si escribe 'fin'
        if entrada == "fin":
            break
            
        try:
            total_visitantes = int(entrada)
            if total_visitantes > 0:
                break
            print("❌ Por favor ingresa un número mayor a 0.")
        except ValueError:
            print("❌ Error: Ingresa un número entero válido o la palabra 'fin'.")

    # Si escribió 'fin', rompe el ciclo principal y pasa al Corte de Caja
    if entrada == "fin":
        print("\nCerrando la taquilla...")
        break

    # Procesar la cantidad de boletos especificada
    for i in range(1, total_visitantes + 1):
        print(f"\n--- Visitante {i} de {total_visitantes} ---")
        
        # Pedir edad con validación
        while True:
            entrada_edad = input("Edad del visitante: ")
            try:
                edad = int(entrada_edad)
                break
            except ValueError:
                print("❌ Error: Ingresa un número entero válido para la edad.")

        # Validación de edad fuera de rango (0 a 110) con continue[cite: 3]
        if edad < 0 or edad > 110:
            print("⚠️ Edad no válida. No se cobra este visitante.")
            continue

        # 4. LÓGICA DE PRECIOS Y DESCUENTOS[cite: 3]
        if edad < 3:
            precio_base = PRECIO_NINO_MENOR
            tipo_texto = "Menor de 3 años"
            descuento = 0.0
            
        elif edad <= 17:
            precio_base = PRECIO_MENOR
            tipo_texto = "Menor de edad (3-17)"
            descuento = 0.0
            
        else:
            # 18 años o más: A TODOS se les pregunta su tipo de visitante obligatoriamente
            precio_base = PRECIO_ADULTO
            
            print("Tipo de visitante: 1) Estudiante  2) Profesor  3) Ninguno")
            tipo_opcion = input("Selecciona una opción (1, 2 o 3): ").strip()
            
            if tipo_opcion == "1":
                tipo_texto = "Estudiante"
                descuento = precio_base * DESC_ESTUDIANTE_PROFESOR
            elif tipo_opcion == "2":
                tipo_texto = "Profesor"
                descuento = precio_base * DESC_ESTUDIANTE_PROFESOR
            else:
                tipo_texto = "Ninguno"
                descuento = 0.0

            # REGLA DE PRIORIDAD PARA ADULTOS MAYORES (60+ años):
            # Se les preguntó su opción, pero se aplica de forma automática el 12% de adulto mayor
            if edad >= 60:
                tipo_texto = f"Adulto mayor ({tipo_texto})"
                descuento = precio_base * DESC_ADULTO_MAYOR

        # 5. CÁLCULOS INDIVIDUALES Y ACUMULACIÓN
        paga_visitante = precio_base - descuento
        
        visitantes_cobrados += 1
        total_descuentos += descuento
        total_pagar += paga_visitante

        # Detalle del visitante actual[cite: 3]
        print(f"👉 Detalle: Edad: {edad} | {tipo_texto} | Base: ${precio_base:.2f} | Descuento: ${descuento:.2f} | A pagar: ${paga_visitante:.2f}")

    # PREGUNTA AL FINAL DE LA COMPRA
    respuesta = input("\n¿Desea comprar otros boletos? (si/no): ").strip().lower()
    
    if respuesta not in ["si", "sí"]:
        print("\nCerrando la taquilla...")
        break

# 6. CORTE DE CAJA FINAL[cite: 3]
print("\n=== CORTE DE CAJA ===")
print(f"Visitantes cobrados: {visitantes_cobrados}")
print(f"Total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_pagar:.2f}")
```
