```python

# PARTE A: TUPLAS

def sumar_tupla(tupla):
    suma = 0

    for numero in tupla:
        suma += numero

    return suma


def seccion_tuplas():
    print("\n--- SECCIÓN 1: TUPLAS ---")

    # Creo una tupla con cinco números
    numeros = (10, 20, 30, 40, 50)

    # Muestro la tupla original y su tercer elemento
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    # Solicito dos números al usuario
    numero1 = int(input("Ingresa el primer número: "))
    numero2 = int(input("Ingresa el segundo número: "))

    # Creo una nueva tupla sin modificar la original
    nueva_tupla = numeros + (numero1, numero2)

    print("Nueva tupla:", nueva_tupla)

    # Convierto la nueva tupla en lista y la ordeno
    lista_ordenada = list(nueva_tupla)
    lista_ordenada.sort()

    print("Lista ordenada:", lista_ordenada)

    # Sumo los elementos mediante la función
    resultado = sumar_tupla(nueva_tupla)

    print("Suma de la nueva tupla:", resultado)
```

Parte A. Tuplas
En esta sección desarrollé un programa para trabajar con tuplas, consultar sus elementos, agregar nuevos números sin modificar la tupla original y realizar operaciones con ellos.
- def sumar_tupla(tupla):: Creé una función que recibe una tupla y se encarga de sumar todos sus elementos.
- suma = 0: Inicialicé una variable para almacenar el resultado.
- for numero in tupla:: Utilicé un ciclo para recorrer cada número de la tupla.
- suma += numero: Fui acumulando los valores para obtener la suma total.
- return suma: Regresé el resultado de la operación.
- numeros = (10, 20, 30, 40, 50): Creé una tupla con cinco números enteros.
- numeros[2]: Consulté el tercer elemento de la tupla, recordando que los índices comienzan desde cero.
- input() e int(): Solicité dos números al usuario y los convertí en enteros.
- numeros + (numero1, numero2): Creé una nueva tupla con los números ingresados, sin modificar la original.
- list(nueva_tupla): Convertí la tupla en una lista para poder ordenar sus elementos.
- lista_ordenada.sort(): Organicé los números de menor a mayor.
- sumar_tupla(nueva_tupla): Llamé a la función para calcular la suma de todos los elementos.
Con esta sección practiqué el uso de tuplas, listas, ciclos, funciones y valores de retorno.

```python

# PARTE B: DICCIONARIOS

def buscar_telefono(contactos, nombre):
    telefono = contactos.get(nombre)
    return telefono


def seccion_diccionarios():
    print("\n--- SECCIÓN 2: DICCIONARIOS ---")

    # Creo un diccionario con tres contactos
    contactos = {
        "Ana": "4421234567",
        "Luis": "4422345678",
        "Maria": "4423456789"
    }

    # Solicito un nuevo contacto
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa el teléfono del nuevo contacto: ")

    # Agrego el nuevo contacto al diccionario
    contactos[nombre_nuevo] = telefono_nuevo

    print("\nContacto agregado correctamente.")

    # Recorro las claves para mostrar solamente los nombres
    print("\nNombres registrados:")

    for nombre in contactos.keys():
        print(nombre)

    # Solicito el nombre que se desea buscar
    nombre_buscar = input("\nIngresa el nombre del contacto que deseas buscar: ")

    # Busco el teléfono utilizando la función
    resultado = buscar_telefono(contactos, nombre_buscar)

    if resultado is not None:
        print("El teléfono de", nombre_buscar, "es:", resultado)
    else:
        print(nombre_buscar, "no está en tus contactos.")
```

Parte B. Diccionarios
En esta sección desarrollé un sistema sencillo para administrar contactos. Mi propósito fue almacenar nombres y teléfonos, agregar nuevos registros y buscar contactos existentes o inexistentes.
- def buscar_telefono(contactos, nombre):: Creé una función que recibe el diccionario y el nombre que deseo buscar.
- contactos.get(nombre): Busqué el nombre dentro del diccionario. Si existe, devuelve su teléfono; si no, devuelve None.
- return telefono: Regresé el resultado de la búsqueda.
- contactos = {...}: Creé un diccionario con tres contactos iniciales, donde cada nombre es una clave y cada teléfono es su valor.
- input(): Solicité el nombre y teléfono del nuevo contacto.
- contactos[nombre_nuevo] = telefono_nuevo: Agregué el nuevo contacto al diccionario.
- contactos.keys(): Obtuve los nombres registrados en el diccionario.
- for nombre in contactos.keys():: Recorrí las claves para mostrar todos los nombres.
- buscar_telefono(contactos, nombre_buscar): Llamé a la función para buscar el contacto solicitado.
- if resultado is not None:: Comprobé si el contacto existe y, si es así, mostré su teléfono.
- else:: Mostré un mensaje cuando el contacto no se encuentra registrado.
Con esta sección aprendí a organizar información mediante diccionarios, consultar sus datos y controlar las búsquedas sin provocar errores cuando un contacto no existe.

```python

# PARTE C: EXCEPCIONES

def seccion_excepciones():
    print("\n--- SECCIÓN 3: EXCEPCIONES ---")

    try:
        # Solicito dos números enteros
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))

        # Realizo las operaciones
        suma = numero1 + numero2
        division = numero1 / numero2

        print("La suma es:", suma)
        print("La división es:", division)

    except ValueError:
        print("Error: debes ingresar números enteros y no dejar campos vacíos.")

    except ZeroDivisionError:
        print("Error: no es posible dividir entre cero.")
```

Parte C. Excepciones
En esta sección realicé operaciones matemáticas con dos números enteros. También incorporé excepciones para controlar los errores que pueden ocurrir cuando el usuario ingresa datos incorrectos.
- def seccion_excepciones():: Definí la función que contiene las operaciones matemáticas.
- try:: Coloqué las instrucciones que podrían generar errores para poder controlarlos.
- input(): Solicité los dos números al usuario.
- int(): Convertí los datos ingresados en números enteros.
- suma = numero1 + numero2: Realicé la suma de los dos valores.
- division = numero1 / numero2: Dividí el primer número entre el segundo.
- print(): Mostré los resultados de ambas operaciones.
- except ValueError:: Capturé los errores que ocurren cuando se ingresan letras o se deja un campo vacío.
- except ZeroDivisionError:: Controlé específicamente el intento de dividir entre cero.
Con esta sección comprendí la importancia de utilizar excepciones para controlar diferentes errores y evitar que el programa termine inesperadamente.

```python

# PARTE D: STRINGS

def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


def seccion_strings():
    print("\n--- SECCIÓN 4: STRINGS ---")

    # Creo una variable de texto
    mensaje = "Aprender programacion mejora nuestras habilidades"

    # Muestro la longitud del mensaje
    print("Mensaje original:", mensaje)
    print("Longitud del mensaje:", len(mensaje))

    # Convierto el mensaje a mayúsculas
    print("Mensaje en mayúsculas:", mensaje.upper())

    # Reemplazo una palabra por otra
    mensaje_modificado = mensaje.replace("programacion", "tecnologia")

    print("Mensaje con reemplazo:", mensaje_modificado)

    # Cuento las palabras utilizando la función
    total_palabras = contar_palabras(mensaje)

    print("Número de palabras:", total_palabras)
```

Parte D. Strings
En esta sección trabajé con cadenas de texto para conocer su longitud, convertirlas a mayúsculas, reemplazar palabras y contar cuántas contiene un mensaje.
- def contar_palabras(texto):: Creé una función que recibe un texto para contar sus palabras.
- texto.split(): Separé el mensaje en palabras utilizando los espacios como referencia.
- return len(palabras): Conté los elementos de la lista y devolví el total.
- mensaje = "Aprender programacion mejora nuestras habilidades": Creé una cadena de texto con cinco palabras.
- len(mensaje): Calculé la cantidad total de caracteres, incluyendo los espacios.
- mensaje.upper(): Convertí todas las letras del mensaje a mayúsculas.
- mensaje.replace("programacion", "tecnologia"): Reemplacé una palabra por otra y guardé el resultado en una nueva variable.
- contar_palabras(mensaje): Llamé a mi función para conocer el número total de palabras.
Con esta sección practiqué diferentes métodos para manipular cadenas de texto y comprendí cómo obtener información de ellas mediante funciones.

```python

# PARTE E: MENÚ MODULAR

def menu_principal():

    while True:
        print("\n========== MENÚ PRINCIPAL ==========")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            seccion_tuplas()

        elif opcion == "2":
            seccion_diccionarios()

        elif opcion == "3":
            seccion_excepciones()

        elif opcion == "4":
            seccion_strings()

        elif opcion == "5":
            print("\nPrograma finalizado. ¡Hasta pronto!")
            break

        else:
            print("\nOpción inválida. Selecciona una opción del 1 al 5.")


# Inicio del programa
menu_principal()
```

Parte E. Menú modular
En esta última sección integré las cuatro funciones anteriores mediante un menú principal. De esta manera, el usuario puede elegir qué sección desea ejecutar y regresar al menú sin reiniciar el programa.
- def menu_principal():: Creé la función que controla el menú principal.
- while True:: Utilicé un ciclo para mantener el menú activo hasta que el usuario decida salir.
- print(): Mostré las cinco opciones disponibles.
- opcion = input(): Solicité al usuario que seleccionara una opción. La guardé como texto para compararla con las alternativas del menú.
- if opcion == "1":: Ejecuté la sección de tuplas.
- elif opcion == "2":: Ejecuté la sección de diccionarios.
- elif opcion == "3":: Ejecuté la sección de excepciones.
- elif opcion == "4":: Ejecuté la sección de strings.
- elif opcion == "5":: Mostré el mensaje de despedida cuando el usuario desea salir.
- break: Finalicé el ciclo while para terminar el programa.
- else:: Mostré un mensaje cuando se ingresa una opción inválida y permití que el menú volviera a aparecer.
- menu_principal(): Llamé a la función para iniciar el programa.
Con esta sección integré todos los temas de la actividad y apliqué la programación modular, organizando cada operación en una función independiente.
