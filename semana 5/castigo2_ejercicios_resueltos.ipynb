## Ejercicio 1 — Bitácora de la semana

```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    indice_mayor = 0
    mayor_km = 0

    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            dias_rodados += 1
        if km[i] > mayor_km:
            mayor_km = km[i]
            indice_mayor = i

    promedio = round(total / dias_rodados, 1) if dias_rodados > 0 else 0
    return total, dias_rodados, promedio, indice_mayor

total, dias_rodados, promedio, indice = resumen(km)
print(f"Total: {total} km")
print(f"Días con rodada: {dias_rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[indice]} ({km[indice]} km)")
```

## Ejercicio 2 — La votación del festival

```python
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

def promedio_votos(lista):
    total = 0
    for banda, cantidad in lista:
        total += cantidad
    return total / len(lista)

ganador = votos[0]
for banda, cantidad in votos:
    if cantidad > ganador[1]:
        ganador = (banda, cantidad)

promedio = promedio_votos(votos)
print(f"Ganador: {ganador[0]} con {ganador[1]} votos")
print(f"Promedio: {promedio}")
print("Arriba del promedio:")
for banda, cantidad in votos:
    if cantidad > promedio:
        print(f"- {banda} ({cantidad})")
```

## Ejercicio 3 — Tramos de la ruta a la Sierra

```python
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

tramo_mayor = ("", "", 0)
for i in range(len(paradas) - 1):
    origen, km_origen = paradas[i]
    destino, km_destino = paradas[i + 1]
    distancia = km_destino - km_origen
    print(f"{origen} → {destino}: {distancia} km")
    if distancia > tramo_mayor[2]:
        tramo_mayor = (origen, destino, distancia)

print(f"Tramo más largo: {tramo_mayor[0]} → {tramo_mayor[1]} ({tramo_mayor[2]} km)")
```

## Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frío"
    elif t <= 24:
        return "templado"
    else:
        return "calor"

etiquetas = []
contador = {"frío": 0, "templado": 0, "calor": 0}

for temperatura in temps:
    etiqueta = clasificar(temperatura)
    etiquetas.append(etiqueta)
    contador[etiqueta] += 1

predomina = ""
mayor_cantidad = 0
for categoria, cantidad in contador.items():
    if cantidad > mayor_cantidad:
        predomina = categoria
        mayor_cantidad = cantidad

print(etiquetas)
print(contador)
print(f"Predomina: {predomina}")
```

## Ejercicio 5 — ¿Cuánto dura el show?

```python
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    return f"{minutos}:{segundos:02d}"

total_segundos = 0
for cancion, duracion in setlist:
    print(f"{cancion}: {formato(duracion)}")
    total_segundos += duracion

print(f"Duración total: {formato(total_segundos)}")
```

## Ejercicio 7 — Boleta del grupo

```python
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for nota in lista:
        suma += nota
    return round(suma / len(lista), 1)

promedios = {}
mejor_alumno = ""
mejor_nota = -1

for alumno, calificaciones in notas.items():
    resultado = promedio(calificaciones)
    promedios[alumno] = resultado
    estado = "Aprobado" if resultado >= 70 else "Reprobado"
    print(f"{alumno}: {resultado} - {estado}")
    if resultado > mejor_nota:
        mejor_nota = resultado
        mejor_alumno = alumno

print(f"Mejor promedio: {mejor_alumno}")
```

## Ejercicio 10 — Traductor de mensajes de WhatsApp

```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    traducidas = []
    for palabra in palabras:
        traducidas.append(abrev.get(palabra, palabra))
    return " ".join(traducidas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

## Ejercicio 13 — Capturas sucias

**Nota:** `int()` acepta espacios al inicio y al final, por eso `" 6 "` se convierte correctamente; `"8.5"` no se puede convertir directamente a entero porque contiene un punto decimal.

```python
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0

for dato in capturas:
    try:
        calificacion = int(dato)
        validas.append(calificacion)
    except ValueError:
        invalidas += 1

promedio = sum(validas) / len(validas) if validas else 0
print(f"Válidas: {validas}")
print(f"Inválidas: {invalidas}")
print(f"Promedio: {round(promedio, 2)}")
```

## Ejercicio 15 — Cupones de descuento

```python
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(total, codigo):
    codigo_limpio = codigo.strip().upper()
    try:
        porcentaje = cupones[codigo_limpio]
        return total - (total * porcentaje / 100)
    except KeyError:
        print("Cupón no válido")
        return total

codigos = ["  moto20 ", "GRATIS", "metal10"]
for codigo in codigos:
    resultado = aplicar(1200, codigo)
    print(f"'{codigo}' → ${resultado}")
```

## Ejercicio 16 — ¿Tu contraseña aguanta?

```python
def evaluar(pw):
    fallas = []
    tiene_numero = False
    tiene_mayuscula = False

    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")

    for caracter in pw:
        if caracter.isdigit():
            tiene_numero = True
        if caracter.isupper():
            tiene_mayuscula = True

    if not tiene_numero:
        fallas.append("falta un número")
    if not tiene_mayuscula:
        fallas.append("falta una mayúscula")
    if " " in pw:
        fallas.append("sin espacios")

    return fallas

contrasenas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]
for pw in contrasenas:
    resultado = evaluar(pw)
    print(f"{pw}: {resultado if resultado else 'segura'}")
```

## Ejercicio 18 — Contador de vocales

```python
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    cantidad = 0
    for caracter in palabra.lower():
        if caracter in "aeiou":
            cantidad += 1
    return cantidad

conteo = {}
for palabra in frase.split():
    conteo[palabra] = vocales(palabra)

palabra_mayor = ""
cantidad_mayor = 0
for palabra, cantidad in conteo.items():
    if cantidad > cantidad_mayor:
        palabra_mayor = palabra
        cantidad_mayor = cantidad

print(conteo)
print(f"Más vocales: {palabra_mayor} ({cantidad_mayor})")
```

## Ejercicio 19 — Usuarios del laboratorio

```python
alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]

usuarios_usados = []

def usuario(nombre):
    partes = nombre.lower().split()
    base = partes[0][0] + partes[-1]
    candidato = base
    numero = 2

    while candidato in usuarios_usados:
        candidato = base + str(numero)
        numero += 1

    usuarios_usados.append(candidato)
    return candidato

for nombre in alumnos:
    print(f"{nombre} → {usuario(nombre)}")
```

## Ejercicio 21 — Estacionamiento del foro

```python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]

def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 1:
        return 20
    return 20 + (horas - 1) * 15

cobros = {}
for placa, entrada, salida in autos:
    horas = salida - entrada
    importe = cobro(entrada, salida)
    cobros[placa] = importe
    print(f"{placa}: {horas} h → ${importe}")

total_dia = 0
for importe in cobros.values():
    total_dia += importe
print(f"Total del día: ${total_dia}")
```

## Ejercicio 22 — Tabla del torneo

```python
partidos = (("Pumas", "Tigres", 2, 1), ("América", "Pumas", 0, 0), ("Tigres", "América", 3, 2),
            ("Pumas", "América", 1, 2), ("Tigres", "Pumas", 1, 1))

puntos = {}
for local, visitante, goles_local, goles_visitante in partidos:
    if local not in puntos:
        puntos[local] = 0
    if visitante not in puntos:
        puntos[visitante] = 0

    if goles_local > goles_visitante:
        puntos[local] += 3
    elif goles_local < goles_visitante:
        puntos[visitante] += 3
    else:
        puntos[local] += 1
        puntos[visitante] += 1

campeon = ""
mayor_puntaje = -1
for equipo, puntaje in puntos.items():
    if puntaje > mayor_puntaje:
        campeon = equipo
        mayor_puntaje = puntaje

print(puntos)
print(f"Campeón: {campeon} con {mayor_puntaje} puntos")
```
